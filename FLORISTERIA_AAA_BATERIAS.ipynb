```python
# Inventario inicial con 'stock'
flores = {
    "tipo": ["margarita", "azucena", "rosa"],
    "stock": [10, 20, 30],
    "precios": [1200, 3000, 1200]
}

# Inicializamos el carrito de compras fuera del ciclo
compra_user = {
    'tipo': [],
    'cantidad': [],
    'precio': []
}

print("Bienvenido a AAA BATERIAS")
print("Se ofrecen los siguientes productos: \n")

# Encabezados de la tabla con stock
print(f"{'Flores':<15} | {'Stock':<10} | {'Precio Unitario':<15}")
print("-" * 46)

# Recorremos las tres listas simultáneamente usando zip
for tipo, stock, precio in zip(flores['tipo'], flores['stock'], flores['precios']):
    print(f"{tipo.capitalize():<15} | {stock:<10} | ${precio:<14}")

while True:
    print("\nQué desea hacer?")
    print(f"{1}. Comprar flores")
    print(f"{2}. Cambiar de flores (Quitar lo comprado)")
    print(f"{3}. Pagar")
    print(f"{4}. Generar factura")
    print(f"{5}. Salir")

    opc = input("Seleccione su opción: \n")
    match opc:
        case "1":
            print("Ha seleccionado: Comprar flores\n")

            # Validar si ya no hay flores
            if len(flores['tipo']) == 0:
                print("¡Lo sentimos! Ya no hay flores, no hay stock disponible.")
                continue

            flor_opc = input("Seleccione la flor que desea comprar: ").lower().strip()
            if flor_opc in flores['tipo']:
                idx = flores['tipo'].index(flor_opc)
                flor_cant = int(input("Ingrese la cantidad que desea comprar: "))

                if flor_cant <= flores['stock'][idx]:
                    # Agregamos los valores a las listas del carrito
                    compra_user['tipo'].append(flor_opc)
                    compra_user['cantidad'].append(flor_cant)
                    compra_user['precio'].append(flores['precios'][idx])

                    # Restamos del stock
                    flores['stock'][idx] -= flor_cant
                    print(f"¡Se han añadido {flor_cant} {flor_opc}(s) al carrito!")

                    # Removemos del inventario las flores si llegan a 0 usando .pop()
                    if flores['stock'][idx] == 0:
                        flores["tipo"].pop(idx)
                        flores["stock"].pop(idx)
                        flores["precios"].pop(idx)

                else:
                    print("Cantidad no disponible en el stock")
            else:
                print(f"La flor {flor_opc} no está disponible o ya no hay stock")

        case "2":
            # Opción para cambiar de flores
            if len(compra_user["tipo"]) > 0:
                print("\nHa decidido cambiar de flores. Se quitarán las que compró y se devolverán al stock.")
                for t, c in zip(compra_user['tipo'], compra_user['cantidad']):
                    if t in flores['tipo']:
                        i_f = flores['tipo'].index(t)
                        flores['stock'][i_f] += c
                    else:
                        flores['tipo'].append(t)
                        flores['stock'].append(c)
                        if t == "margarita": p_orig = 1200
                        elif t == "azucena": p_orig = 3000
                        else: p_orig = 1200
                        flores['precios'].append(p_orig)

                # Vaciamos el carrito del usuario
                compra_user = {'tipo': [], 'cantidad': [], 'precio': []}
                print("¡Se han quitado las flores del carrito y devuelto al stock con éxito!")
            else:
                print("El carrito de compras ya está vacío.")

        case "3":
            if len(compra_user["tipo"]) > 0:
                print("Seleccione el medio de pago: \n")
                print(f"{1}. Efectivo")
                print(f"{2}. Tarjeta de crédito")
                print(f"{3}. Tarjeta de débito")
                print(f"{4}. Otro medio de pago")
                pago_opc = input("Seleccione su opción: \n")

                # Calculamos el total multiplicando cada cantidad por su precio unitario
                total_pagar = 0
                for i in range(len(compra_user["tipo"])):
                    subtotal = compra_user["cantidad"][i] * compra_user["precio"][i]
                    total_pagar += subtotal

                print(f"\nUsted ha seleccionado la opción de pago {pago_opc}.")
                print(f"El total a pagar por su compra es: ${total_pagar}")

            else:
                print("El carrito de compras está vacío. ")

        case "4":
            print("Ha seleccionado: Generar factura\n")
            if len(compra_user["tipo"]) > 0:
                # Encabezados de la tabla con alineación mostrando la cantidad comprada
                print(f"{'Flores':<15} | {'Cantidad':<10} | {'Precio Unitario':<15} | {'Subtotal':<15}")
                print("-" * 60)
                total_factura = 0
                for tipo, cantidad, precio in zip(compra_user['tipo'], compra_user['cantidad'], compra_user['precio']):
                    subtotal = precio * cantidad
                    total_factura += subtotal
                    print(f"{tipo.capitalize():<15} | {cantidad:<10} | ${precio:<14} | ${subtotal:<14}")
                print("-" * 60)
                print(f"TOTAL A PAGAR: ${total_factura}")
            else:
                print("El carrito está vacío, no se puede generar factura.")
            break

        case "5":
            print("Ha seleccionado: Salir\n")
            break
        case _:
            print('Número incorrecto, por favor seleccione uno correcto')

print("Gracias por emplear nuestros servicios en AAA BATERIAS. Feliz día")
```
