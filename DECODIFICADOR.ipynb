```python
def codificar_palabra(palabra):
  resultado = []
  for char in palabra:
    if "a" <= char <= "z":
      nuevo_char = chr((ord(char) - ord("a") + 1) % 26 + ord("a"))
      resultado.append(nuevo_char)
    elif "A" <= char <= "Z":
      nuevo_char = chr((ord(char) - ord("A") + 1) % 26 + ord("A"))
      resultado.append(nuevo_char)
    else:
      resultado.append(char)
  return "".join(resultado)
entrada = input("Introduce la palabra: ")

salida = codificar_palabra(entrada)

print(f"\nPalabra original: {entrada}")
print(f"Palabra codificada: {salida}")
```
