```python
import math
import ipywidgets as widgets
import matplotlib.pyplot as plt
from IPython.display import display, clear_output

l1 = widgets.FloatText(value=10, description="Eslabón 1:")
l2 = widgets.FloatText(value=8, description="Eslabón 2:")
ang1 = widgets.FloatText(value=45, description="Ángulo 1:")
ang2 = widgets.FloatText(value=30, description="Ángulo 2:")
boton = widgets.Button(description="Calcular")
salida = widgets.Output()
def calcular(b):
    with salida:
        clear_output()
        r1 = math.radians(ang1.value)
        r2 = math.radians(ang2.value)
        codo_x = l1.value * math.cos(r1)
        codo_y = l1.value * math.sin(r1)
        pos_x = codo_x + l2.value * math.cos(r1 + r2)
        pos_y = codo_y + l2.value * math.sin(r1 + r2)
        print(f"Eje X: {pos_x:.2f} cm")
        print(f"Eje Y: {pos_y:.2f} cm")
        archivo = open("resultados.txt", "a")
        archivo.write(f"L1={l1.value} L2={l2.value} A1={ang1.value} A2={ang2.value} -> X={pos_x:.2f} Y={pos_y:.2f}\n")
        archivo.close()
        plt.plot([0, codo_x, pos_x], [0, codo_y, pos_y], "o-", linewidth=4)
        plt.grid()
        plt.axis("equal")
        plt.title("Posición del brazo")
        plt.show()
boton.on_click(calcular)
display(l1, l2, ang1, ang2, boton, salida)
```
