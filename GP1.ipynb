```python
import random
print("Bienvenido al juego de Piedra Papel o Tijera")
opciones = ["PIEDRA", "PAPEL", "TIJERA"]
r = int(input("Quieres jugar?? escribe el número de la respuesta 1.SI // 2.NO 1"))
if r == 1:
  print("Estas jugando contra una computadora, prueba tu suerte")
  mano = int(input("Ingrese el número de lo que quiera sacar. 1.PIEDRA // 2.PAPEL // 3.TIJERA"))
  if mano in [1, 2, 3]:
        jugador = opciones[mano - 1]
        pc = random.choice(opciones)
        print(f"Tú elegiste: {jugador}")
        print(f"La computadora eligió: {pc}")
        if jugador == pc:
          print("Estan empatados")
        elif (jugador == "PIEDRA" and pc == "TIJERA") or (jugador == "PAPEL" and pc == "PIEDRA") or (jugador == "TIJERA" and pc == "PAPEL"):
          print("FELICIDADES, HAS LOGRADO DERROTAR A LA COMPUTADORA!!!!!! :) ")
        else:
          print("PERDISTE, LA COMPUTADORA PREDIJO TUS MOVIMIENTOS... ")
  else:
    print("Opción no valida, debes escoger entre el 1, 2 o 3")
else:
  print("Esta bien tu te lo pierdes :( ")
```
