#DICCIONARIO
Un diccionario es una coleccion de elementos donde cada uno tiene una key o llave y asociada tiene un value o valor.

Propiedades:
* Dinamicos: pueden crecer y añadir elementos.
* Son indexados desde las llaves
* Pueden ser anidados


```
diccionario={}
diccionario={llave:valor,llave2:valor2,...,llaven:valorn}
```
